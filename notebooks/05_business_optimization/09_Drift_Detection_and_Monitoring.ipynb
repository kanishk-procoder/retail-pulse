### Observability Scope:
- Monitors covariate data drift between reference baseline (2009-2010) and current inference (2011).
- Evaluates Kolmogorov-Smirnov (KS) two-sample test.
- Automated retraining alert triggers when p-value < 0.05.

```python
import os
import pandas as pd
import numpy as np
from scipy.stats import ks_2samp
import matplotlib.pyplot as plt

data_path = os.path.join('..', '..', 'data', 'processed', 'cleaned_transactions.parquet')
df = pd.read_parquet(data_path)
df['Date'] = pd.to_datetime(df['Date'])

ref_data = df[df['Date'] < '2011-06-01']['TotalAmount']
curr_data = df[df['Date'] >= '2011-06-01']['TotalAmount']

stat, p_val = ks_2samp(ref_data, curr_data)
print(f'Kolmogorov-Smirnov Statistic: {stat:.4f}')
print(f'Drift Test p-value:          {p_val:.5f}')
if p_val < 0.05:
    print('ALERT: Statistically significant covariate drift detected. Trigger retraining pipeline recommendation.')
else:
    print('STATUS: No significant covariate drift detected. Model weights remain valid.')
```
