### Objective:
- Pre-aggregate all KPIs, trends, country breakdowns, and predictions into `dashboard/data/`.
- Ensures sub-second rendering across all Streamlit views.

```python
import os, json
import pandas as pd

kpi_path = os.path.join('..', '..', 'dashboard', 'data', 'kpis.json')
with open(kpi_path) as f: kpis = json.load(f)

print('=== PRODUCTION DASHBOARD KPI SUMMARY ===')
for k, v in kpis.items():
    if isinstance(v, float):
        print(f'  {k:28s}: {v:,.2f}')
    else:
        print(f'  {k:28s}: {v:,}')
```
