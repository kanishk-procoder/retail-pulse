# 04e: Customer Segmentation Model Benchmarking & Champion Selection

**RetailPulse AI Platform** • *Comparative Clustering Evaluation*

### Objective:
- Load benchmark metrics across all 4 clustering models.
- Compare Silhouette Score, Calinski-Harabasz, and Davies-Bouldin.
- Formally declare and promote Champion Model.

```python
import os, json
import pandas as pd
import matplotlib.pyplot as plt

metrics_path = os.path.join('..', '..', 'reports', 'metrics', 'segmentation_metrics.json')
with open(metrics_path) as f: metrics = json.load(f)

df_m = pd.DataFrame(metrics)
print('=== CUSTOMER SEGMENTATION MASTER LEADERBOARD ===')
print(df_m[['Model', 'Clusters', 'Silhouette_Score', 'Calinski_Harabasz', 'Davies_Bouldin', 'Interpretability', 'Selected']].to_string(index=False))
```

```python
# Side-by-Side Metric Comparison Plots
fig, axes = plt.subplots(1, 3, figsize=(18, 4))

axes[0].bar(df_m['Model'], df_m['Silhouette_Score'], color='#3B82F6')
axes[0].set_title('Silhouette Score (Higher is Better)', fontweight='bold')
axes[0].tick_params(axis='x', rotation=30)

axes[1].bar(df_m['Model'], df_m['Calinski_Harabasz'], color='#10B981')
axes[1].set_title('Calinski-Harabasz Index (Higher is Better)', fontweight='bold')
axes[1].tick_params(axis='x', rotation=30)

axes[2].bar(df_m['Model'], df_m['Davies_Bouldin'], color='#F59E0B')
axes[2].set_title('Davies-Bouldin Index (Lower is Better)', fontweight='bold')
axes[2].tick_params(axis='x', rotation=30)

plt.tight_layout()
plt.show()

print('\nCHAMPION SELECTION: K-Means achieves highest Silhouette (0.3425), highest Calinski-Harabasz (4,848.23), and best business interpretability.')
```
